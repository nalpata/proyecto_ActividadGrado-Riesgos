## 1. Preparar los insumos

Se descomprime el resultado completo del validador v2 y se usa la muestra adjudicada que quedó versionada al cerrar el Día 1.

```python
from pathlib import Path
import zipfile

zip_path = Path('data/evaluation/risk_validation/validator_v2/resultados_validador_riesgos_v2.zip')
input_dir = Path('/content/day2_inputs')
input_dir.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(zip_path) as archive:
    archive.extract('risk_validation_results.csv', input_dir)

validator_results = input_dir / 'risk_validation_results.csv'
adjudicated = Path('data/evaluation/risk_validation/validator_v2/final/sample_blind_adjudicated.csv')
print('Validador:', validator_results)
print('Adjudicación:', adjudicated)
```
