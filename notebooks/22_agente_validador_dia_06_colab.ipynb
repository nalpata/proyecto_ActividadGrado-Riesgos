## 1. Subir el resultado calibrado del Día 5
Selecciona `resultados_clasificador_calibrado_dia_05.zip`.

```python
from google.colab import files
from pathlib import Path
import zipfile, shutil
uploaded=files.upload()
zip_name=next(n for n in uploaded if n.startswith('resultados_clasificador_calibrado_dia_05'))
source=Path('/content/day5_calibrado'); source.mkdir(exist_ok=True)
with zipfile.ZipFile(zip_name) as z: z.extractall(source)
input_csv=source/'documentary_items_calibrated.csv'
assert input_csv.exists(), 'El ZIP no contiene documentary_items_calibrated.csv'
print('Archivo validado:', input_csv)
```

## 2. Ejecutar el agente validador

```python
output=Path('/content/resultados_dia_06'); output.mkdir(exist_ok=True)
!python -m src.risk.run_validation_agent --input "{input_csv}" --output-dir "{output}"
```

## 3. Verificar y descargar

```python
import json, pandas as pd
summary=json.loads((output/'validation_agent_summary.json').read_text())
print(json.dumps(summary,ensure_ascii=False,indent=2))
assert summary['items_received']==1049
assert summary['items_accepted']+summary['items_rejected']==1049
zip_path=shutil.make_archive('/content/resultados_agente_validador_dia_06','zip',output)
files.download(zip_path)
```
